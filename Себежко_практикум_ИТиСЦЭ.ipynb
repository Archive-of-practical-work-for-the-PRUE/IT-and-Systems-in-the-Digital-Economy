###Задача 4.2. Визуализация данных в виде графиков.

####Подзадача 1. Визуализация логистических функций.

```python
import numpy as np
import matplotlib.pyplot as plt

# Задаем начальные параметры
variant_student = 1
w = 13.05
step = 0.007

# Создаем массив значений x
x = np.arange(-w * variant_student, w * variant_student, step)

# Создаем графики
plt.figure(figsize=(5, 10))

# 1. Логистическая функция
plt.subplot(3, 1, 1)
plt.plot(x, 1 / (1 + np.exp(-x)), label='signoid', color='blue', linestyle='-')
plt.title(f'Вариант № {variant_student}')
plt.xlabel('Себежко & Кузнецов')
plt.ylabel('Группа ПИ1/23б')
plt.legend()

# 2. Гиперболический тангенс
plt.subplot(3, 1, 1)
plt.plot(x, np.tanh(x), label='tanh', color="y", linestyle='--')
plt.legend()

# 3. ReLU
plt.subplot(3, 1, 2)
plt.plot(x, np.maximum(0, x), label='ReLU', color=np.random.rand(3,), linestyle='-.')
plt.title(f'Вариант № {variant_student}')
plt.xlabel('Себежко & Кузнецов')
plt.ylabel('Группа ПИ1/23б')
plt.legend()

# Отображаем графики
plt.tight_layout() # отступы
plt.show()
```

###Задача 4.3. Нейронные сети.

####Подзадача 1. Реализация на Python задачи 3.1.

```python
import numpy as np
from datetime import datetime

# Задаем начальные параметры
input_layer = [1, -1, 3, 2]
output_layer = [1, 1.5]
weights = [[0.5, -0.2, -0.4, -0.1], [0.7, -0.1, -0.1, -0.9]]

# Находим сумму приходящих весов
V1 = sum(x * y for x, y in zip(input_layer, weights[0]))
V2 = sum(x * y for x, y in zip(input_layer, weights[1]))

# Находим значение нейронов V1 и V2
value_V1 = 1 / (1 + np.exp(-V1))
value_V2 = 1 / (1 + np.exp(-V2))

# Находим ошибку сети
error_net = ((output_layer[0] - value_V1)**2 + (output_layer[1] - value_V2)**2) / 2

# Вывод результатов
print("Задача 4.3. Нейронные сети")
print("ПОДЗАДАЧА 1. Реализация на Python задачи 3.1")
print("Студент: Себежко Александр Андреевич, Группа ПИ1/23б, Вариант 1")
print(f"Дата выполнения работы: {datetime.now()}\n")

print(f"Входной слой: {input_layer}")
print(f"Выходной слой: {output_layer}")
print(f"Веса: {weights}\n")

print(f"Суммарное входное воздействие для нейрона V1: {V1:.6f} Значение нейрона V1: {value_V1:.6f}")
print(f"Суммарное входное воздействие для нейрона V2: {V2:.6f} Значение нейрона V2: {value_V2:.6f}\n")

print(f"Ошибка сети: {error_net:.6f}")
```

####Подзадача 2. Реализация на Python задачи 3.3 Backpropagation.

```python
# Задаем начальные параметры
input_layer, output_layer = [1, -1, 3, 2], [1, 1.5]
weights = [[0.5, -0.2, -0.4, -0.1], [0.7, -0.1, -0.1, -0.9]]

# Находим сумму приходящих весов
V1 = sum(x * y for x, y in zip(input_layer, weights[0]))
V2 = sum(x * y for x, y in zip(input_layer, weights[1]))

# Находим значение нейронов V1 и V2
value_V1 = 1 / (1 + np.exp(-V1))
value_V2 = 1 / (1 + np.exp(-V2))

# Находим ошибку сети
error_net = ((output_layer[0] - value_V1)**2 + (output_layer[1] - value_V2)**2) / 2

# Вывод результатов 0 итерации
print("Задача 4.3. Нейронные сети")
print("ПОДЗАДАЧА 1. Реализация на Python задачи 3.1")
print("Студент: Себежко Александр Андреевич, Группа ПИ1/23б, Вариант 1")
print(f"Дата выполнения работы: {datetime.now()}\n")

print(f"Входной слой: {input_layer}")
print(f"Выходной слой: {output_layer}")
print(f"Веса: {weights}\n")

print(f"Первоначально рассчитанное значение нейрона V1: {value_V1:.6f}")
print(f"Первоначально рассчитанное значение нейрона V2: {value_V2:.6f}\n")

print(f"Первоначально рассчитанная ошибка: {error_net:.6f}\n")

errors = [error_net] # Массив с ошибками
iterations = 100 # введите кол-во желаемых итераций
iterations += 1

for i in range(1, iterations):
  # Находим производную
  f_V1 = value_V1 * (1 - value_V1)
  f_V2 = value_V2 * (1 - value_V2)

  # Находим дельту
  delta_v1 = f_V1 * error_net
  delta_v2 = f_V2 * error_net

  # Обновляем веса
  weights = [
    [round(float(w + n * delta_v1 * 0.11), 6) for w, n in zip(weights[0], input_layer)],
    [round(float(w + n * delta_v2 * 0.11), 6) for w, n in zip(weights[1], input_layer)]
  ]

  # Находим сумму приходящих весов
  V1 = sum(n * w for n, w in zip(input_layer, weights[0]))
  V2 = sum(n * w for n, w in zip(input_layer, weights[1]))

  # Находим значение нейронов V1 и V2
  value_V1 = 1 / (1 + np.exp(-V1))
  value_V2 = 1 / (1 + np.exp(-V2))

  # Находим ошибку сети
  error_net = ((output_layer[0] - value_V1)**2 + (output_layer[1] - value_V2)**2) / 2

  print(f"Новые веса на ( {i} ) итерации: {weights}")
  print(f"Ошибка сети ( {i} ): {error_net:.6f}\n")

  errors.append(error_net) # Добавляем ошибку в список

plt.figure(figsize=(8, 5))
plt.plot(range(iterations), errors, linestyle='-', color='royalblue', linewidth=2.5)
plt.title("Себежко & Кузнецов, группа ПИ1/23б, Вариант № 1")
plt.xlabel('Цикл обучения', fontsize=10)
plt.ylabel('Ошибка сети', fontsize=10)
plt.grid(False)
plt.xticks(range(0, iterations, 20))
plt.show()
```
